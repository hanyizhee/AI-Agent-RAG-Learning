```python
import pandas as pd

# 构造DataFrame -- 创建数据集（学员成绩信息）
df = pd.DataFrame([
    {"姓名":"小王","语文":90,"数学":89,"英语":80},
    {"姓名":"小孙","语文":93,"数学":88,"英语":82},
    {"姓名":"小钱","语文":94,"数学":92,"英语":88},
    {"姓名":"小周","语文":78,"数学":90,"英语":92},
    {"姓名":"小吴","语文":90,"数学":88,"英语":81},
    {"姓名":"小郑","语文":80,"数学":65,"英语":80},
    {"姓名":"小韩","语文":96,"数学":86,"英语":95},
])

# 统计计算
print(f"语文最高分是{df['语文'].max()},语文最低分是{df['语文'].min()},语文平均分是{df['语文'].mean():.2f}")
print(f"数学最高分是{df['数学'].max()},数学最低分是{df['数学'].min()},数学平均分是{df['数学'].mean():.2f}")
print(f"英语最高分是{df['英语'].max()},语文最低分是{df['英语'].min()},英语平均分是{df['英语'].mean():.2f}")

```
