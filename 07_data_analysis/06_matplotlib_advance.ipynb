```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt

# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei'] # 将中文设置为黑体

# 创建子图
# figure:画布对象,axes:子图数组
figure,axes = plt.subplots(nrows=1,ncols=2,figsize=(10,5),dpi=100) # nrows:行,ncols:列

# 图一：柱状图
countries = ['美国','中国','英国','日本','德国']# 国家列表
values = [19.5, 20.2, 17.3, 16.5, 15.2]

axes1:Axes = axes[0]
axes1.bar(countries,values,color='green',width=0.6)

axes1.set_xlabel('国家')
axes1.set_ylabel('石油储量(亿吨)')
axes1.grid(linestyle='--',alpha=0.3)
axes1.set_title('世界石油储存储备',fontsize=18)

# 图二：饼状图 --> 世界人口（擅长比例分析）
countries1 = ['美国','中国','英国','日本','德国','法国','俄罗斯']
values1 = [19.5, 20.2, 17.3, 16.5, 15.2, 14.5, 13.2]

axes2:Axes = axes[1]
axes2.pie(values1,labels=countries1,autopct='%1.1f%%') # 前后的%为转义字符
axes2.set_title('世界人口比例',fontsize=18)
axes2.legend(loc='lower center',ncol=5,bbox_to_anchor=(0.5,-0.1)) # 前面的值控制的水平方位，后面的值控制的垂直方位

# 保存图片
plt.savefig('data/01.png')


plt.show()
```
